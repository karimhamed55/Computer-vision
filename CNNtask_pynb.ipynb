```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix, classification_report

keras.utils.set_random_seed(42)
AUTOTUNE = tf.data.AUTOTUNE

for gpu in tf.config.list_physical_devices("GPU"):
    try:
        tf.config.experimental.set_memory_growth(gpu, True)
    except RuntimeError:
        pass

# DATASET
(x_train_full, y_train_full), (x_test_full, y_test_full) = keras.datasets.fashion_mnist.load_data()
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat", "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]

x_train, _, y_train, _ = train_test_split(
    x_train_full, y_train_full, train_size=0.5, random_state=42, stratify=y_train_full
)
x_test, _, y_test, _ = train_test_split(
    x_test_full, y_test_full, train_size=0.5, random_state=42, stratify=y_test_full
)

x_train, x_val, y_train, y_val = train_test_split(
    x_train, y_train, test_size=0.2, random_state=42, stratify=y_train
)

print("Dataset: Fashion-MNIST")
print("Using 50% of the original dataset")
print("Training:", x_train.shape, " | Validation:", x_val.shape, " | Testing:", x_test.shape)

plt.figure(figsize=(10, 3))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(class_names[y_train[i]])
    plt.axis("off")
plt.tight_layout()
plt.show()

# CNN FROM SCRATCH
cnn_train_ds = (
    tf.data.Dataset.from_tensor_slices((x_train, y_train))
    .shuffle(10000, seed=42)
    .batch(128)
    .prefetch(AUTOTUNE)
)
cnn_val_ds = tf.data.Dataset.from_tensor_slices((x_val, y_val)).batch(128).prefetch(AUTOTUNE)

cnn = keras.Sequential([
    keras.Input(shape=(28, 28, 1)),
    layers.Rescaling(1.0 / 255),
    layers.RandomTranslation(0.1, 0.1),
    layers.Conv2D(32, 3, padding="same", activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),
    layers.Dropout(0.2),
    layers.Conv2D(64, 3, padding="same", activation="relu"),
    layers.BatchNormalization(),
    layers.MaxPooling2D(),
    layers.Dropout(0.3),
    layers.Conv2D(128, 3, padding="same", activation="relu"),
    layers.BatchNormalization(),
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.4),
    layers.Dense(10, activation="softmax")
])

cnn.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
cnn.summary()

start = time.time()
history_cnn = cnn.fit(
    cnn_train_ds,
    validation_data=cnn_val_ds,
    epochs=8,
    callbacks=[
        keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=2, restore_best_weights=True),
        keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=1, min_lr=1e-5)
    ],
    verbose=1
)
cnn_training_time = time.time() - start

cnn_probabilities = cnn.predict(x_test, batch_size=256, verbose=0)
cnn_predictions = np.argmax(cnn_probabilities, axis=1)
cnn_accuracy = accuracy_score(y_test, cnn_predictions)
cnn_precision, cnn_recall, cnn_f1, _ = precision_recall_fscore_support(
    y_test, cnn_predictions, average="weighted", zero_division=0
)

# PRETRAINED EFFICIENTNETB0
IMAGE_SIZE = 96
EFF_BATCH_SIZE = 64

def resize_for_efficientnet(image, label):
    image = tf.cast(image, tf.float32)
    image = tf.expand_dims(image, axis=-1)
    image = tf.image.grayscale_to_rgb(image)
    image = tf.image.resize(image, (IMAGE_SIZE, IMAGE_SIZE))
    return image, label

def resize_image_only(image):
    image = tf.cast(image, tf.float32)
    image = tf.expand_dims(image, axis=-1)
    image = tf.image.grayscale_to_rgb(image)
    return tf.image.resize(image, (IMAGE_SIZE, IMAGE_SIZE))

eff_train_ds = (
    tf.data.Dataset.from_tensor_slices((x_train, y_train))
    .shuffle(10000, seed=42)
    .map(resize_for_efficientnet, num_parallel_calls=AUTOTUNE)
    .batch(EFF_BATCH_SIZE)
    .prefetch(AUTOTUNE)
)
eff_val_ds = (
    tf.data.Dataset.from_tensor_slices((x_val, y_val))
    .map(resize_for_efficientnet, num_parallel_calls=AUTOTUNE)
    .batch(EFF_BATCH_SIZE)
    .prefetch(AUTOTUNE)
)
eff_test_ds = (
    tf.data.Dataset.from_tensor_slices(x_test)
    .map(resize_image_only, num_parallel_calls=AUTOTUNE)
    .batch(EFF_BATCH_SIZE)
    .prefetch(AUTOTUNE)
)
eff_test_labeled_ds = (
    tf.data.Dataset.from_tensor_slices((x_test, y_test))
    .map(resize_for_efficientnet, num_parallel_calls=AUTOTUNE)
    .batch(EFF_BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

base_model = keras.applications.EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(IMAGE_SIZE, IMAGE_SIZE, 3)
)
base_model.trainable = False

eff_inputs = keras.Input(shape=(IMAGE_SIZE, IMAGE_SIZE, 3))
eff_x = base_model(eff_inputs, training=False)
eff_x = layers.GlobalAveragePooling2D()(eff_x)
eff_x = layers.Dropout(0.3)(eff_x)
eff_outputs = layers.Dense(10, activation="softmax")(eff_x)
efficientnet = keras.Model(eff_inputs, eff_outputs)

efficientnet.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
efficientnet.summary()

start = time.time()
history_eff_head = efficientnet.fit(
    eff_train_ds,
    validation_data=eff_val_ds,
    epochs=3,
    callbacks=[keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=1, restore_best_weights=True)],
    verbose=1
)

base_model.trainable = True
for layer in base_model.layers[:-20]:
    layer.trainable = False
for layer in base_model.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

efficientnet.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history_eff_fine = efficientnet.fit(
    eff_train_ds,
    validation_data=eff_val_ds,
    epochs=2,
    callbacks=[keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=1, restore_best_weights=True)],
    verbose=1
)
efficientnet_training_time = time.time() - start

eff_probabilities = efficientnet.predict(eff_test_ds, verbose=0)
eff_predictions = np.argmax(eff_probabilities, axis=1)
eff_accuracy = accuracy_score(y_test, eff_predictions)
eff_precision, eff_recall, eff_f1, _ = precision_recall_fscore_support(
    y_test, eff_predictions, average="weighted", zero_division=0
)

# RESULTS
results = pd.DataFrame({
    "Metric": ["Test accuracy", "Weighted precision", "Weighted recall", "Weighted F1-score", "Training time (minutes)", "Parameters"],
    "CNN from scratch": [cnn_accuracy, cnn_precision, cnn_recall, cnn_f1, cnn_training_time / 60, cnn.count_params()],
    "EfficientNetB0": [eff_accuracy, eff_precision, eff_recall, eff_f1, efficientnet_training_time / 60, efficientnet.count_params()]
})

print("\nFINAL MODEL COMPARISON")
print(results.to_string(index=False, formatters={
    "CNN from scratch": lambda value: f"{value:.4f}" if value < 100 else f"{value:,.0f}",
    "EfficientNetB0": lambda value: f"{value:.4f}" if value < 100 else f"{value:,.0f}"
}))

metric_names = ["Accuracy", "Precision", "Recall", "F1-score"]
cnn_scores = [cnn_accuracy, cnn_precision, cnn_recall, cnn_f1]
eff_scores = [eff_accuracy, eff_precision, eff_recall, eff_f1]
positions = np.arange(len(metric_names))
width = 0.35

plt.figure(figsize=(9, 5))
plt.bar(positions - width / 2, cnn_scores, width, label="CNN from scratch")
plt.bar(positions + width / 2, eff_scores, width, label="EfficientNetB0")
plt.xticks(positions, metric_names)
plt.ylim(0, 1)
plt.ylabel("Score")
plt.title("CNN vs EfficientNetB0 - Fashion-MNIST")
plt.legend()
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 5))
for i, (predictions, title) in enumerate([
    (cnn_predictions, "CNN from scratch"),
    (eff_predictions, "EfficientNetB0")
]):
    cm = confusion_matrix(y_test, predictions)
    plt.subplot(1, 2, i + 1)
    plt.imshow(cm, cmap="Blues")
    plt.title(title)
    plt.xlabel("Predicted label")
    plt.ylabel("True label")
    plt.xticks(range(10), class_names, rotation=90)
    plt.yticks(range(10), class_names)
    plt.colorbar(fraction=0.046, pad=0.04)
plt.tight_layout()
plt.show()

print("\nCNN CLASSIFICATION REPORT")
print(classification_report(y_test, cnn_predictions, target_names=class_names, zero_division=0))
print("EFFICIENTNETB0 CLASSIFICATION REPORT")
print(classification_report(y_test, eff_predictions, target_names=class_names, zero_division=0))

if eff_accuracy > cnn_accuracy:
    print(f"\nConclusion: EfficientNetB0 achieved the higher test accuracy ({eff_accuracy:.4f}) compared with the CNN trained from scratch ({cnn_accuracy:.4f}).")
elif cnn_accuracy > eff_accuracy:
    print(f"\nConclusion: The CNN trained from scratch achieved the higher test accuracy ({cnn_accuracy:.4f}) compared with EfficientNetB0 ({eff_accuracy:.4f}).")
else:
    print(f"\nConclusion: Both models achieved the same test accuracy ({cnn_accuracy:.4f}).")
```
